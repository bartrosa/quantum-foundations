```python
# w notebooku albo w Python REPL
from quantum_foundations.conway_games.causet import make_diamond_conway
from quantum_foundations.conway_games.automorphism import aut_order_conway
from quantum_foundations.conway_games.entropy import (
    twin_classes_canonical, twin_classes_local
)
import math

c = make_diamond_conway()  # 4 events
print(f"n={c.n}, ranks={c.rank}")
print(f"P_L = {c.past_left}")
print(f"P_R = {c.past_right}")

aut = aut_order_conway(c, backend="networkx")
canon = twin_classes_canonical(c)
local = twin_classes_local(c)

prod_canon = math.prod(math.factorial(len(cls)) for cls in canon)
prod_local = math.prod(math.factorial(len(cls)) for cls in local)

print(f"|Aut| = {aut}")
print(f"canonical classes = {canon}, prod_k! = {prod_canon}")
print(f"local classes = {local}, prod_k! = {prod_local}")
print(f"local lemma: {aut} = ?·{prod_local}, ratio = {aut/prod_local}")
print(f"canon inequality: {aut} <= ?·{prod_canon}, slack = {prod_canon} - {aut}")
```
