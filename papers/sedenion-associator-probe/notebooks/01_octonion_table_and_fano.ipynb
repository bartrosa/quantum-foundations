# Octonion multiplication table & Fano lines

**Paper:** Appendix (`tab:oct`), §Preliminaries — quaternion triples span Fano lines.

Run Jupyter from the **repository root** (`quantum-foundations/`) after `uv sync --all-extras`, so the kernel sees the editable install of `quantum_foundations`.

**Claims exercised:** incidence of the seven lines of $\mathrm{PG}(2,2)$ encoded in `OCT_TAB`; each line $\{i,j,k\}$ satisfies $e_ie_j \propto e_k$.

```python
from quantum_foundations.sedenion.algebra import OCT_TAB
from quantum_foundations.sedenion.fano import FANO_LINES, compute_fano_lines

assert len(FANO_LINES) == 7
assert FANO_LINES == compute_fano_lines()

# Each point 1..7 lies on exactly three lines (dual of PG(2,2))
for p in range(1, 8):
    on = sum(1 for ln in FANO_LINES if p in ln)
    assert on == 3, p

print("Fano lines (as triples of indices):")
for ln in sorted(FANO_LINES, key=lambda s: tuple(sorted(s))):
    print(" ", tuple(sorted(ln)))
```

```python
# Row e1 from OCT_TAB matches paper appendix (e1·e1 = -e0 → (-1,0), etc.)
assert OCT_TAB[1][0] == (1, 1)
assert OCT_TAB[1][1] == (-1, 0)
assert OCT_TAB[1][2] == (1, 3)

# Lines from products e_a e_b ∝ e_k match FANO_LINES construction
from itertools import combinations

seen = set()
for a in range(1, 8):
    for b in range(a + 1, 8):
        _sgn, k = OCT_TAB[a][b]
        seen.add(frozenset({a, b, k}))
assert seen == set(FANO_LINES)
print("OCT_TAB-derived triples coincide with FANO_LINES.")
```

**Result:** Table `OCT_TAB` reproduces the appendix signs and yields exactly seven incidence triples — the Fano plane — as in Lemma 3.2 / incidence preliminaries.
